# Data Cleaning — Spam Email Classification

```python
import sys
sys.path.append('../src')
import pandas as pd
from preprocessing import clean_text

df = pd.read_csv('../data/raw/emails.csv')
print('Before:', df.shape)
```

## Remove duplicate records

```python
before = df.shape[0]
df = df.drop_duplicates().reset_index(drop=True)
print(f'Removed {before - df.shape[0]} duplicate rows')
```

## Normalize text
Lowercase, strip the repeated 'Subject:' prefix, and collapse whitespace, so the TF-IDF vectorizer sees consistent tokens instead of noisy variants.

```python
df['text'] = df['text'].apply(clean_text)
```

## Drop any rows left empty after cleaning

```python
before = df.shape[0]
df = df[df['text'].str.len() > 0].reset_index(drop=True)
print(f'Removed {before - df.shape[0]} empty rows')
print('Final shape:', df.shape)
```

```python
df.to_csv('../data/processed/emails_cleaned.csv', index=False)
df['spam'].value_counts()
```
